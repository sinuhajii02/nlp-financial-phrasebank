## Split Dataset
Split `all-data.csv` into three sets:
- **Train**: used to fit the models.
- **Val**: used to tune and compare models.
- **Test**: held out until the end for a final, unbiased evaluation.


```python
from sklearn.model_selection import train_test_split
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("../data/all-data.csv", header=None, names=['label', 'sentence'], encoding='latin-1')
df.head()
```

```python
print(df.shape)
print(df.describe())
print(df["label"].isna().sum())
```

```python
print("Counts: ")
print(df['label'].value_counts(), "\n")
print("Proportion")
share_label = df['label'].value_counts(normalize=True).round(3)
print(share_label.multiply(100))
```

For an imbalanced dataset like this, we use macro F1 as our main metric, as it exposes a model that ignores negatives.

```python
n_before = len(df)
df = df.drop_duplicates()
print("Found exact duplicates: ", n_before - len(df))
```

```python
duplicate = df[df.duplicated("sentence", keep=False)]
print(display(duplicate))
df = df.drop_duplicates("sentence", keep=False)
print(len(df))
```

```python
train_val, test = train_test_split(df, test_size=0.2, random_state=42, stratify=df['label'])

print(len(train_val))

```

```python
train, val = train_test_split(train_val, test_size=0.25, random_state=42, stratify=train_val['label'])
```

```python
print("Train: ", len(train))
print("Val: ", len(val))
print("Test: ", len(test))
print("Total equal: ", (len(train) + len(val) + len(test)) == len(df))
```

### Save to file

```python
train.to_csv("../data/train.csv", index=False)
val.to_csv("../data/val.csv", index=False)
test.to_csv("../data/test.csv", index=False)
```
